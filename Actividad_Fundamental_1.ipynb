# ***Ejercicio 1: Evaluador de Rendimiento de Modelos***

Lógica implementada: Se solicita al usuario ingresar el valor de exactitud (accuracy) del modelo, validando que sea un número decimal dentro del rango de 0.0 a 1.0. Mediante estructuras condicionales (if, elif y else), se clasifica el rendimiento del modelo en cuatro categorías: Excelente, Aceptable, Requiere Ajuste o Crítico/No apto. Se utiliza un ciclo while para repetir la solicitud en caso de errores y break para finalizar cuando se introduce un valor válido.

```python
def evaluar_modelo():
  while True:
    try:
      op=float(input("introduzca el valor de la exactitud de su modelo (un valor entre el 0.0 y el 1.0): "))
      if op>=0 and op <=1:
        if op>=0.90:
          print("Categoria de su modelo: Excelente")
        elif op>=0.75:
          print("Categoria de su modelo: Aceptable")
        elif op>=0.50:
          print("Categoria de su modelo: Requiere Ajuste")
        else:
          print("Categoria de su modelo: Crítico / No apto")
        break
      else:
        print("el valor introducido no es valido")
    except:
      print("tipo de dato introducido no valido")

evaluar_modelo()
```

# ***Ejercicio 2: Procesamiento de Hiperparametros y Lotes***

Lógica implementada: Se define una configuración global con los valores (512, 10) y una función que recibe una lista de tasas de aprendizaje. Mediante un ciclo for, se recorre cada elemento de la lista y se multiplica por un factor de escala de 2, almacenando los resultados en una nueva lista. Finalmente, la función devuelve la lista modificada junto con la configuración global. Se ejecuta la función con cuatro tasas de aprendizaje diferentes para mostrar su procesamiento.

```python
global CONFIGURACION_BASE
CONFIGURACION_BASE=(512,10)
```

```python
def procesar_tasas_aprendizaje(lista_tasas=[]):
  f_escala=2
  nueva_lista=[]
  for i in lista_tasas:
    nueva_lista.append(i*f_escala)
  return nueva_lista, CONFIGURACION_BASE
```

# ***Ejercicio 3: Simulador de Convergencia por Epocas***

Lógica implementada: Se solicita al usuario ingresar el número de épocas que desea simular, validando que sea un número entero positivo. Se utiliza un ciclo while junto con try-except para controlar errores en la entrada de datos. Posteriormente, mediante un ciclo for, se simula el entrenamiento calculando una pérdida (loss) decreciente en cada época. Se utiliza una condición if para comprobar si la pérdida es menor a 0.05; si se cumple, se indica que el modelo ha convergido anticipadamente y se detiene el ciclo mediante break.

```python
def simular_entrenamiento():
  while True:
    try:
      op=int(input("introduzca el numero de epocas a simular (numero entero positivo): "))
      if op>=0:
        for i in range(1,op+1,1):
          loss = 1 / i

          print("Epoca:", i, "| Loss:", round(loss, 4))

          if loss < 0.05:
              print("El modelo ha convergido anticipadamente.")
              break
        break
      else:
        print("el valor introducido no es valido")
    except:
      print("tipo de dato introducido no valido")
simular_entrenamiento()
```
